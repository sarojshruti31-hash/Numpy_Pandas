```python
# series
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

#index
print(s[0])
print(s[1])

#label
s = pd.Series([23, 24, 25, 26], index = ["Adam", "Eve", "bob", "Charlie"])
print(s, type(s))
print(s.index)
```

```python
# Properties of series

#Vectorized operations are allowed
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1 + s2)
s1[0] = 25
changed_s1 = s1.drop(0)
print(s1)
print(changed_s1)
```

```python
# DataFrame (Dictionary of lists)
info = {
    "Name" : ["Rahul", "Harshita", "Shraddha"],
    "CGPA" : [9.5, 9.0, 8.5],
    "Age" : [20, 21, 19]
}
df = pd.DataFrame(info)
print(df, type(df))
print(df.index)
print(df.columns)
```

```python
df
```

```python
# List os Lists
df = pd.DataFrame([["Adam", 23], ["Shruti", 21], ["Lucky", 21]], columns = ["Name", "Age"])
print(df, type(df))
```

```python
df
```

```python
# np array
import numpy as np
np_arr = np.array([[1,2,3], [4,5,6]])
df = pd.DataFrame(np_arr, columns = ["A", "B", "C"]) 
print(df, type(df))
```
